```python
from groq import Groq
from dotenv import load_dotenv
import os
```

```python
load_dotenv()
```

```python
key = os.getenv("GROQ_API_KEY")
```

```python
client = Groq(api_key=key)
```

```python
chatHistory = []
```

```python
def ask_llm(user_query):
    chatHistory.append({ "role":"user",
                "content":user_query})

    
    message = chatHistory

    chat_completion = client.chat.completions.create(messages=message,model = "openai/gpt-oss-120b")
    response = chat_completion.choices[0].message.content

    chatHistory.append({ "role":"assistant",
                    "content":response})
    return response
```

```python
ask_llm("Whats my name")
```

```python
history = []

while True:
    user_query = input("Enter query")

    if(user_query.lower()=="exit"):
        break

    
    message = {
        "role":"user",
        "content":user_query
    }

    history.append(message)

    chat_completion = client.chat.completions.create(messages=history,model = "openai/gpt-oss-120b")
    response = chat_completion.choices[0].message.content
    
    history.append({ "role":"assistant",
                        "content":response})

    print(response)
```

```python
def ask_llm_temp1(prompt):
    chat_completion = client.chat.completions.create(
        messages=[
            {
                "role": "user",
                "content": prompt,
            }
        ],

        # The language model which will generate the completion.
        model="openai/gpt-oss-120b",
        temperature=1
        )
    response = chat_completion.choices[0].message.content

    return response
```

```python
query = "Create an story about python in 5 points"
print(ask_llm_temp1(query))
```

```python
def ask_llm_temp_token(user_query,max_token = 1000,temperature = 0):
    message = [
        {
            "role":"user",
            "content":user_query
        }
    ]
    chat_completion = client.chat.completions.create( messages=message,model="openai/gpt-oss-20b",temperature=temperature,max_tokens=max_token)
    response = chat_completion.choices[0].message.content
    return response
```

<h2>Few Shot with Placeholder</h2>

```python
prompts = """Classify the sentiment as Positive,Negative,Netural, Reply with label only
Review : Movie is Awesome
Sentiment : Positive
Review : Movie was bad
Sentiment : Negative
Review : {review}
Sentiment"""
```

```python
input_query = prompts.format(review="I loved the movie")
```

```python
print(ask_llm(input_query))
```

```python
reviews = ["I loved the movie","I hate the movie","I am confused about movie story"]

for r in reviews:
    print(r)
    print(ask_llm(prompts.format(review=r)))
```
